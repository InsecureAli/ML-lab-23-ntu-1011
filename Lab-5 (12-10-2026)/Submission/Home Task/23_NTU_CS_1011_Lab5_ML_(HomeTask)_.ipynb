# `SVM`

````python
# Import all relevant libraries
from sklearn.svm import SVC
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.metrics import accuracy_score, confusion_matrix
import warnings
warnings.filterwarnings("ignore")

# Load dataset
df = pd.read_csv('income_evaluation.csv')

# Strip whitespace from column names if any
df.columns = df.columns.str.strip()
df['income'] = df['income'].str.strip()

# Identify categorical and numerical features
categorical_features = df.select_dtypes(include=['object']).columns.drop('income')
numerical_features = df.select_dtypes(exclude=['object']).columns

# Apply one-hot encoding to categorical features
df_encoded = pd.get_dummies(df, columns=categorical_features, drop_first=True)

# Separate features and target variable
X = df_encoded.drop("income", axis=1)
y = (df_encoded["income"] == ">50K").astype(int)

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=101)

# Scale numerical features
scaler = StandardScaler()
X_train[numerical_features] = scaler.fit_transform(X_train[numerical_features])
X_test[numerical_features] = scaler.transform(X_test[numerical_features])

# Train SVM model and check cross-validation scores
svc = SVC(random_state=101)
svc.fit(X_train, y_train)

accuracies = cross_val_score(svc, X_train, y_train, cv=5)
print("Train Score:", np.mean(accuracies))
print("Test Score:", svc.score(X_test, y_test))

# Hyperparameter Tuning using GridSearchCV (Optional)
grid = {
    'C': [0.01, 0.1, 1, 10],
    'kernel': ["linear", "poly", "rbf", "sigmoid"],
    'degree': [1, 3, 5, 7],
    'gamma': [0.01, 1]
}

svm = SVC()
svm_cv = GridSearchCV(svm, grid, cv=5)
svm_cv.fit(X_train, y_train)

print("Best Parameters:", svm_cv.best_params_)
print("Tuned Train Score:", svm_cv.best_score_)
print("Tuned Test Score:", svm_cv.score(X_test, y_test))
````

# ** K-Mean clustering**


````python
# 1. Sab zaroori libraries import karna
from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import pandas as pd
import warnings
warnings.filterwarnings("ignore")

# 2. Dataset load karna
df = pd.read_csv('mushroomss.csv')

# Column names aur target variable ko clean karna
df.columns = df.columns.str.strip()
df['income'] = df['income'].str.strip()

# 3. Categorical aur Numerical features ko alag karna
categorical_features = df.select_dtypes(include=['object']).columns.drop('income')
numerical_features = df.select_dtypes(exclude=['object']).columns

# One-hot encoding apply karna categorical features par
df_encoded = pd.get_dummies(df, columns=categorical_features, drop_first=True)

# Features (X) aur Target (y) alag karna
X = df_encoded.drop("income", axis=1)
y = (df_encoded["income"] == ">50K").astype(int)

# 4. Data ko training aur testing sets mein split karna
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=101)

# 5. Numerical features ko scale karna
scaler = StandardScaler()
X_train[numerical_features] = scaler.fit_transform(X_train[numerical_features])
X_test[numerical_features] = scaler.transform(X_test[numerical_features])

# 6. Base Classifiers initialize karna
log_clf = LogisticRegression(random_state=101)
rnd_clf = RandomForestClassifier(random_state=101)
svm_clf = SVC(random_state=101)

# 7. Voting Classifier banana (Hard voting ke sath)
voting_clf = VotingClassifier(
    estimators=[('lr', log_clf), ('rf', rnd_clf), ('svc', svm_clf)],
    voting='hard'
)

# 8. Sab classifiers ko train karke unki accuracy compare karna
for clf in (log_clf, rnd_clf, svm_clf, voting_clf):
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    print(clf.__class__.__name__, accuracy_score(y_test, y_pred))
#```[cite: 14, 16, 17, 21, 22]
````

# Voting Classifier

````python
# 1. Sab zaroori libraries import karna
from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import pandas as pd
import warnings
warnings.filterwarnings("ignore")

# 2. Mall Customers dataset load karna
df = pd.read_csv('Mall_Customers.csv')

# 3. Supervised classification ke liye target variable banana
# (Misal ke tor par: Agar Spending Score > 50 hai toh 1 (High Spender), warna 0)
df['High_Spender'] = (df['Spending Score (1-100)'] > 50).astype(int)

# Features (X) aur Target (y) alag karna
X = df[['Age', 'Annual Income (k$)']]
y = df['High_Spender']

# 4. Data ko training aur testing sets mein split karna
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=101)

# 5. Features ko scale karna
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 6. Base Classifiers initialize karna
log_clf = LogisticRegression(random_state=101)
rnd_clf = RandomForestClassifier(random_state=101)
svm_clf = SVC(random_state=101)

# 7. Voting Classifier banana (Hard voting ke sath)
voting_clf = VotingClassifier(
    estimators=[('lr', log_clf), ('rf', rnd_clf), ('svc', svm_clf)],
    voting='hard'
)

# 8. Sab classifiers ko train karke unki accuracy compare karna
for clf in (log_clf, rnd_clf, svm_clf, voting_clf):
    clf.fit(X_train_scaled, y_train)
    y_pred = clf.predict(X_test_scaled)
    print(clf.__class__.__name__, accuracy_score(y_test, y_pred))
````
